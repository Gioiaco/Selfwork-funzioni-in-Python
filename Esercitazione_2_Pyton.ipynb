Creare una classe Person , estenderla con due sottoclassi Dipendente e Studente.

```python
#Creare una classe Person , estenderla con due sottoclassi Dipendente e Studente.

class Person:
    
    def __init__(self, name, age):
        self.name = name
        self.age = age 

class Dipendente(Person):
    
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary

class Studente(Person):
    
    def __init__(self, name, age, grade):
        super().__init__(name, age)
        self.grade = grade      


dipendente = Dipendente("Mario Rossi", 30, 3000)
studente = Studente("Luca Bianchi", 20, "A")

#  stampare le loro informazioni
print(f"Dipendente: {dipendente.name}, Età: {dipendente.age}, Stipendio: {dipendente.salary}")
print(f"Studente: {studente.name}, Età: {studente.age}, Voto: {studente.grade}")        
```

Estendere Dipendenti con 2 sottoclassi Docente e Freelance


```python
#Estendere Dipendenti con 2 sottoclassi Docente e Freelance


class Docente(Dipendente):
   
    def __init__(self, name, age, salary, subject):
        super().__init__(name, age, salary)
        self.subject = subject

class Freelance(Dipendente):
    
    def __init__(self, name, age, salary, project):
        super().__init__(name, age, salary)
       
        self.project = project  


docente = Docente("Giovanni Verdi", 45, 4000, "Matematica")
freelance = Freelance("Anna Neri", 35, 3500, "Progetto A") 

# stampare le loro informazioni

print(f"Docente: {docente.name}, Età: {docente.age}, Stipendio: {docente.salary}, Materia: {docente.subject}")
print(f"Freelance: {freelance.name}, Età: {freelance.age}, Stipendio: {freelance.salary}, Progetto: {freelance.project}")       
```
